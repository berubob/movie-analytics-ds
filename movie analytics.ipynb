### Data Type Conversion

There are some missed align data type like &darr;
<br>

- ``budget`` and ``popularity`` are ``str`` instead of ``float64`` or ``int64``
<br>

- ``release_data`` is ``str`` instead of ``datetime``

```python
movies_data.dtypes
```

- Converting ``release_date`` data type from ``str`` to ``datetime64[us]``

```python
movies_data[["budget", "popularity"]] = movies_data[["budget", "popularity"]].apply(pd.to_numeric, errors='coerce')
```

- Converting ``budget`` and ``popularity`` from ``str`` to ``float64``

```python
movies_data["release_date"] = pd.to_datetime(movies_data["release_date"], format="%Y-%m-%d", errors='coerce')
```
